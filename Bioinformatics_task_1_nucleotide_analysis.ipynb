```python
# 1. Define the DNA sequence
dna = 'ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA'

# 2. Count each nucleotide individually
count_A = dna.count('A')
count_T = dna.count('T')
count_G = dna.count('G')
count_C = dna.count('C')

# Print individual base counts
print(f'A: {count_A}')
print(f'T: {count_T}')
print(f'G: {count_G}')
print(f'C: {count_C}')

# 3. Calculate GC Content percentage
# GC content is the percentage of nitrogenous bases in a DNA sequence that are either guanine (G) or cytosine (C).
gc = (count_G + count_C) / len(dna) * 100
print(f'GC Content: {gc:.2f}%')

# 4. Count using a Dictionary Comprehension (more concise Python approach)
counts = {base: dna.count(base) for base in 'ATGC'}
print("Counts Dictionary:", counts)
```

```python
import matplotlib.pyplot as plt

# Data from our dictionary
bases = list(counts.keys())
values = list(counts.values())

# Define custom colors for nitrogenous bases
colors = ['#4CAF50', '#2196F3', '#FF9800', '#F44336']

# Create the bar chart
plt.figure(figsize=(7, 5))
plt.bar(bases, values, color=colors, edgecolor='black', width=0.5)

# Add titles and labels
plt.title('Nucleotide Frequency in DNA Sequence', fontsize=14, fontweight='bold')
plt.xlabel('Nucleotide Base', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the plot inside Colab
plt.show()
```
